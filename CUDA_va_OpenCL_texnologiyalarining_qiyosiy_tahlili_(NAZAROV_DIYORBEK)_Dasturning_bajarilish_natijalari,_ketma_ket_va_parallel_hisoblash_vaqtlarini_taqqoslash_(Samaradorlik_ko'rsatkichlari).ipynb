```python
import matplotlib.pyplot as plt

labels = ['Markaziy protsessor (CPU)', 'Grafik protsessor (GPU - CUDA)']
times = [8.30700, 0.00062]
colors = ['#ff9999', '#66b3ff']

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, times, color=colors, edgecolor='black')

for bar, time in zip(bars, times):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.1,
             f'{time} sek', ha='center', va='bottom', fontweight='bold', fontsize=12)

plt.title("CPU va GPU (CUDA) hisoblash vaqtlarini taqqoslash\n(10 millionta element, kichikroq bo'lgani yaxshi)", fontsize=14, pad=15)
plt.ylabel("Bajarilish vaqti (soniya)", fontsize=12)
plt.yscale('log')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
