```python
class PipelineSimulator:
    def __init__(self, r1_value):
        self.registers = {'R1': r1_value, 'R7': 0}
        self.instruction = "DIVI R7, R1, 4"
        self.clock_cycle = 0
        self.history = []

    def log_stage(self, stage, desc):
        self.clock_cycle += 1
        self.history.append((self.clock_cycle, stage, desc))

    def IF_stage(self):
        self.log_stage("IF", f"Buyruq olindi: {self.instruction}")

    def ID_stage(self):
        self.log_stage("ID", f"Dekodlandi. R1 dan o'qildi: {self.registers['R1']}")

    def EX_stage(self):
        self.temp_result = self.registers['R1'] // 4
        self.log_stage("EX", f"Bajarildi: {self.registers['R1']} // 4 = {self.temp_result}")

    def MEM_stage(self):
        self.log_stage("MEM", "Xotiraga murojaat qilinmadi")

    def WB_stage(self):
        self.registers['R7'] = self.temp_result
        self.log_stage("WB", f"Natija yozildi. R7 = {self.registers['R7']}")

    def render_graphics(self):
        print("\n╔══════╦═════╦════════════════════════════════════════════════════╗")
        print("║ TAKT ║ ID  ║ JARAYON TAFSILOTI                                  ║")
        print("╠══════╬═════╬════════════════════════════════════════════════════╣")
        for cycle, stage, desc in self.history:
            print(f"║  {cycle:<3} ║ {stage:<3} ║ {desc:<50} ║")
        print("╚══════╩═════╩════════════════════════════════════════════════════╝")
        print(f"\n   >> YAKUNIY REGISTRLAR: [ R1 = {self.registers['R1']} | R7 = {self.registers['R7']} ] <<\n")

simulator = PipelineSimulator(20)

simulator.IF_stage()
simulator.ID_stage()
simulator.EX_stage()
simulator.MEM_stage()
simulator.WB_stage()

simulator.render_graphics()
```
