```python
from graphviz import Digraph
from IPython.display import display

dot_cpu = Digraph(comment='CPU Algoritmi')

dot_cpu.attr(rankdir='TB', size='4,6')
dot_cpu.attr('node', fontname='Arial', fontsize='10')
dot_cpu.attr('edge', fontname='Arial', fontsize='9')

dot_cpu.node('A', 'Boshlash', shape='oval', style='filled', fillcolor='lightblue')
dot_cpu.node('B', 'N = 10 000 000\nA, B, C massivlarni yaratish', shape='parallelogram', style='filled', fillcolor='lightyellow')
dot_cpu.node('C', 'Sikl: i = 0 dan N gacha', shape='hexagon', style='filled', fillcolor='lightgreen')
dot_cpu.node('D', 'C[i] = A[i] + B[i]', shape='box', style='filled', fillcolor='white')
dot_cpu.node('E', 'Vaqtni o\'lchash va\nnatijani chiqarish', shape='parallelogram', style='filled', fillcolor='lightyellow')
dot_cpu.node('F', 'Tamom', shape='oval', style='filled', fillcolor='lightblue')

dot_cpu.edge('A', 'B')
dot_cpu.edge('B', 'C')
dot_cpu.edge('C', 'D', label=' Shart bajarilsa (i<N)')
dot_cpu.edge('D', 'C', label=' i++')
dot_cpu.edge('C', 'E', label=' Sikl tugasa (i=N)')
dot_cpu.edge('E', 'F')

print("CPU algoritmining blok-sxemasi:")
display(dot_cpu)

dot_gpu = Digraph(comment='GPU Algoritmi')

dot_gpu.attr(rankdir='TB', size='4,8')
dot_gpu.attr('node', fontname='Arial', fontsize='10')
dot_gpu.attr('edge', fontname='Arial', fontsize='9')

dot_gpu.node('A', 'Boshlash', shape='oval', style='filled', fillcolor='lightblue')
dot_gpu.node('B', 'Host (CPU) xotirasida\nmassivlarni yaratish', shape='parallelogram', style='filled', fillcolor='lightyellow')
dot_gpu.node('C', 'Massivlarni Device (GPU)\nxotirasiga ko\'chirish\n(cuda.to_device)', shape='box', style='filled', fillcolor='lightgrey')
dot_gpu.node('D', 'Grid va Block\no\'lchamlarini belgilash', shape='box', style='filled', fillcolor='white')
dot_gpu.node('E', 'CUDA Kernelni parallel ishga tushirish\n(Minglab oqimlarda)', shape='hexagon', style='filled', fillcolor='lightgreen')
dot_gpu.node('F', 'idx = threadIdx.x + ...\nC[idx] = A[idx] + B[idx]', shape='box', style='filled', fillcolor='white')
dot_gpu.node('G', 'Natijani GPU dan CPU ga\nqaytarish (copy_to_host)', shape='box', style='filled', fillcolor='lightgrey')
dot_gpu.node('H', 'Samaradorlikni (Speedup)\nhisoblash va chiqarish', shape='parallelogram', style='filled', fillcolor='lightyellow')
dot_gpu.node('I', 'Tamom', shape='oval', style='filled', fillcolor='lightblue')

dot_gpu.edge('A', 'B')
dot_gpu.edge('B', 'C')
dot_gpu.edge('C', 'D')
dot_gpu.edge('D', 'E')
dot_gpu.edge('E', 'F', label=' Har bir oqim (thread) uchun')
dot_gpu.edge('F', 'G', label=' Barcha oqimlar tugagach (Sync)')
dot_gpu.edge('G', 'H')
dot_gpu.edge('H', 'I')

print("\nGPU (CUDA) algoritmining blok-sxemasi:")
display(dot_gpu)
```
